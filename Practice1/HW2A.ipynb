## 1. Что делает `df.info()`?

```
df.info()
```

Эта команда показывает общую информацию о нашем датасете.

Представь, что у нас есть Excel-таблица с информацией о пассажирах Титаника.

`df.info()` помогает понять:

- Сколько пассажиров в таблице?
- Сколько столбцов?
- Какие типы данных хранятся?
- Есть ли пропущенные значения?
- Нужно ли очищать данные перед обучением модели?

Например, вот эти строки:

```
Age       714 non-null    float64
Cabin     204 non-null    object
Embarked  889 non-null    object
```

Они означают:

| Столбец  | Заполнено | Пропущено |
| -------- | --------- | --------- |
| Age      | 714       | 177       |
| Cabin    | 204       | 687       |
| Embarked | 889       | 2         |

Почему это важно?

У нас 891 пассажир, но возраст известен только для 714. Если мы сразу начнём обучать модель, некоторые алгоритмы не смогут работать с пропущенными значениями.

Поэтому сначала нужно подготовить данные.

## 3. Что означают типы данных?

В твоём выводе есть три типа:

| Тип       | Значение      | Пример         |
| --------- | ------------- | -------------- |
| `int64`   | Целое число   | `Survived = 1` |
| `float64` | Дробное число | `Age = 22.5`   |
| `object`  | Обычно текст  | `Sex = male`   |

Например:

```
df["Age"]       # float64df["Survived"]  # int64df["Sex"]       # object
```

Модель машинного обучения обычно ожидает числовые признаки. Поэтому текстовые значения, например `male` и `female`, потребуется преобразовать в числа.
